```python
import os
from openai import OpenAI

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY")) # I have Encrypted my API Key

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "Give me 3 app ideas using OpenAI APIs"}
    ]
)

print(response.choices[0].message.content)
```
